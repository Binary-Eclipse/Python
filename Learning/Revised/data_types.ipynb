## Password strength validation. 

```python
import re
def password_strength(password):
     pattern = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"

     if re.match(pattern,password):
          return "Stregth Password"
     return "Weak Password"

print(password_strength("weka"))
print(password_strength("weka12!@"))



print("String method: ")
#string method
def check_password_strength(password):
    feedback = []
    
    if len(password) < 8:
        feedback.append("Password must be at least 8 characters long.")
    if not any(char.isupper() for char in password):
        feedback.append("Password must contain at least one uppercase letter.")
    if not any(char.islower() for char in password):
        feedback.append("Password must contain at least one lowercase letter.")
    if not any(char.isdigit() for char in password):
        feedback.append("Password must contain at least one number.")
    
    # Check for special characters using a set intersection
    special_chars = set("!@#$%^&*()-+?_=,<>/")
    if not any(char in special_chars for char in password):
        feedback.append("Password must contain at least one special character.")
        
    # If the feedback list is empty, the password is strong
    if not feedback:
        return True, "Password is strong."
    
    return False, feedback

# Test
is_valid, messages = check_password_strength("Python24")

if not is_valid:
    for msg in messages:
        print(f"- {msg}")
```

## Date validity checker. 

```python
from datetime import date
from datetime import datetime
#not perfect one
def datetime_check(year,month,day):

      try:
        date(year,month,day)
        return "Valid"
      except ValueError:
        return"Invalid"
        

print(datetime_check(3,10,31))
 
# Correct one
def daetetime_valid_string(date_string,date_format="%Y-%m-%d"):
   try:
      datetime.strptime(date_string,date_format)
      return "Valid"
   except ValueError:
      return "Invalid"

print(daetetime_valid_string("2026-10-01"))
```

## Rock-paper-scissors decision logic

```python
def play_rps(p1,p2):
    p1,p2=p1.lower(),p2.lower()
    if p1==p2:
        return "!Its a tie"

    defeate={
       "rock" : "scissors",
       "scissors" : "paper",
       "paper" : "rock"
    }
    if defeate.get(p1)==p2:
        return "Player1 win"
    else:
        return "player 2 win"

print(play_rps("rock","paper"))
```

## Multiple-condition access-control system. 

```python
## way 1
def access_control_system(user):
    condition=[
        user.get("is_active")==True,
        user.get("role") in ["admin","manager"],
        user.get("age",0) >= 18,
        not user.get("is_locked")  # if lock = true, return false means rejected or lock= false ,return true that he can enter
            ]
    if all(condition):
       return True
    return False

user={"is_active":True, "role":"admin", "age":19,"is_locked":False}
print(access_control_system(user))


## better way with feedback
def validate_access(user, required_clearance):
    # 1. Check account status
    if not user.get("is_active"):
        return False, "Access Denied: Account is deactivated."
        
    if user.get("is_locked"):
        return False, "Access Denied: Account is temporarily locked."

    # 2. Check role/clearance level
    if user.get("clearance_level", 0) < required_clearance:
        return False, f"Access Denied: Requires clearance level {required_clearance}."

    # 3. Check MFA (Multi-Factor Authentication)
    if not user.get("mfa_enabled"):
        return False, "Access Denied: Multi-Factor Authentication is required."

    # If it survives all checks, access is granted
    return True, "Access Granted."

# Test
user_3 = {"is_active": True, "is_locked": False, "clearance_level": 2, "mfa_enabled": False}

is_allowed, message = validate_access(user_3, required_clearance=3)
print(message)
```
