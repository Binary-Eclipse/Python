## find maximum

````python
age=[12,14,12,122,21,67]
max_value=float("-inf")
for i in range(len(age)):
    if age[i]>max_value:
        max_value=age[i]

print(max_value)

# max()
print(max(age)) # O(n)
````

## find minimum

````python
age=[12,14,12,122,21,67]
max_value=float("inf")
for i in range(len(age)):
    if age[i]<max_value:
        max_value=age[i]

print(max_value)

#min()
print(min(age))  # O(n)
````

## find sum

````python
marks=[80,67,89,90,76,69]
sums=0
for i in range(len(marks)):
    sums+=marks[i]

print(sums)

# sum()
print(sum(marks))
````

## count occurances

````python
age=[12,34,12,56,34,12,89,56]
target=12
c=0
for i in range (len(age)):
    if target==age[i]:
        c+=1
print(f"{target} Occured {c} times")

# using dictionary
freq={}
for i in range(len(age)):
    freq[age[i]]=freq.get(age[i],0)+1

print(freq)
````

## Rotate list

````python
age=[12,34,12,56,34,12,89,56]

# left rotate by 1
first=age[0]

for i in range (len(age)-1):# 0,1,2,3,4,5,6
     age[i]=age[i+1]

age[-1]=first
print(age)


# right rotate by 1
age=[12,34,12,56,34,12,89,56]
last=age[-1]

for i in range (len(age)-1,0,-1):#7 6 5 4 3 2 1 
     age[i]=age[i-1]

age[0]=last
print(age)
````

## Find missing number.
```
-Approch1: if value is start from 1 to n
  then, find expected = (n*n+1)//2 
  and missing=expected-actual

-Approach2: if value start from randomly but have consequitive
such as: 64,65,67,68

````python
#approach 1
numbers = [1, 2, 3, 5, 6]
n=len(numbers)+1 # added 1 bcz 1 value is missing
actual=0
expected=(n*(n+1))//2
for i in numbers:
    actual+=i

print(f"missing:{expected-actual}")
````

````python
#approach 2
numbers = [64, 65, 66, 67, 69,71]
for i in range(len(numbers)-1):
    if numbers[i+1]-numbers[i] > 1 :
        print(f"missing value: {numbers[i]+1}")
````

## Move zeros to end.

````python
#approach 1
l1=[2,0,3,4,5,0,6]
pos=0
for i in range(len(l1)):
    if l1[i] != 0:
       l1[pos]=l1[i]
       pos+=1
while pos<len(l1):
    l1[pos]=0
    pos+=1
print(l1)
````

## Find pair with target sum.
- adjacent version
- any pair sum = target 

````python
#adjacent
l1=[2,4,9,10,3]
target=13
for i in range(1,len(l1)):
    if l1[i-1]+l1[i]==target:
        print(f"pair{l1[i-1],l1[i]}")
````

````python
#any pair
l1=[2,4,9,10,3,4]
target=13
for i in range(len(l1)):
    for j in range(i+1,len(l1)):
        if l1[i]+l1[j] == target:
            print(f"pair{l1[i],l1[j]}")
````

## Find triplets with target sum.

````python
l1=[2,4,9,10,3,4]
target=15
for i in range (len(l1)):
    for j in range(i+1,len(l1)):
        for k in range(j+1,len(l1)):
            if l1[i]+l1[j]+l1[k]==target:
                print(f"triplets{l1[i],l1[j],l1[k]}")
````
