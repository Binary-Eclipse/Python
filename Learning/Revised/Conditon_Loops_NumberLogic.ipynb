## Password strength validation. 

```python
import re
def password_strength(password):
     pattern = r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"

     if re.match(pattern,password):
          return "Stregth Password"
     return "Weak Password"

print(password_strength("weka"))
print(password_strength("weka12!@"))



print("String method: ")
#string method
def check_password_strength(password):
    feedback = []
    
    if len(password) < 8:
        feedback.append("Password must be at least 8 characters long.")
    if not any(char.isupper() for char in password):
        feedback.append("Password must contain at least one uppercase letter.")
    if not any(char.islower() for char in password):
        feedback.append("Password must contain at least one lowercase letter.")
    if not any(char.isdigit() for char in password):
        feedback.append("Password must contain at least one number.")
    
    # Check for special characters using a set intersection
    special_chars = set("!@#$%^&*()-+?_=,<>/")
    if not any(char in special_chars for char in password):
        feedback.append("Password must contain at least one special character.")
        
    # If the feedback list is empty, the password is strong
    if not feedback:
        return True, "Password is strong."
    
    return False, feedback

# Test
is_valid, messages = check_password_strength("Python24")

if not is_valid:
    for msg in messages:
        print(f"- {msg}")
```

## Date validity checker. 

```python
from datetime import date
from datetime import datetime
#not perfect one
def datetime_check(year,month,day):

      try:
        date(year,month,day)
        return "Valid"
      except ValueError:
        return"Invalid"
        

print(datetime_check(3,10,31))
 
# Correct one
def daetetime_valid_string(date_string,date_format="%Y-%m-%d"):
   try:
      datetime.strptime(date_string,date_format)
      return "Valid"
   except ValueError:
      return "Invalid"

print(daetetime_valid_string("2026-10-01"))
```

## Rock-paper-scissors decision logic

```python
def play_rps(p1,p2):
    p1,p2=p1.lower(),p2.lower()
    if p1==p2:
        return "!Its a tie"

    defeate={
       "rock" : "scissors",
       "scissors" : "paper",
       "paper" : "rock"
    }
    if defeate.get(p1)==p2:
        return "Player1 win"
    else:
        return "player 2 win"

print(play_rps("scissors","paper"))
```

## Multiple-condition access-control system. 

```python
## way 1
def access_control_system(user):
    condition=[
        user.get("is_active")==True,
        user.get("role") in ["admin","manager"],
        user.get("age",0) >= 18,
        not user.get("is_locked")  # if lock = true, return false means rejected or lock= false ,return true that he can enter
            ]
    if all(condition):
       return True
    return False

user={"is_active":True, "role":"admin", "age":19,"is_locked":False}
print(access_control_system(user))


## better way with feedback
def validate_access(user, required_clearance):
    # 1. Check account status
    if not user.get("is_active"):
        return False, "Access Denied: Account is deactivated."
        
    if user.get("is_locked"):
        return False, "Access Denied: Account is temporarily locked."

    # 2. Check role/clearance level
    if user.get("clearance_level", 0) < required_clearance:
        return False, f"Access Denied: Requires clearance level {required_clearance}."

    # 3. Check MFA (Multi-Factor Authentication)
    if not user.get("mfa_enabled"):
        return False, "Access Denied: Multi-Factor Authentication is required."

    # If it survives all checks, access is granted
    return True, "Access Granted."

# Test
user_3 = {"is_active": True, "is_locked": False, "clearance_level": 2, "mfa_enabled": False}

is_allowed, message = validate_access(user_3, required_clearance=3)
print(message)
```

## Palindrome Number

```python
# string palindrome
def is_palindrome(s):
    reversed=s[::-1]
    return s==reversed

s=input("Enter a string to check if it's a palindrome: ")
print(is_palindrome(s))

num=121
#string palindrome
reverse=int(str(num)[::-1])
if reverse==num:
    print("palindrome")


# numeric palindrome
def is_numeric_palindrome(num):
    reverse=0
    temp=num
    while temp>0:
        digit=temp%10
        reverse=reverse*10+digit
        temp//=10
    return reverse==num

num=121
print(is_numeric_palindrome(num))
```

## Armstrong in range

```python
start=int(input("Enter the start number: "))
ends=int(input("Enter the end number: "))

for num in range(start,ends+1):
    digits=len(str(num))

    power=[i**digits for i in range(10)]

    temp=num
    total=0

    while temp>0:
        digit=temp%10
        total+=power[digit]
        temp//=10

    if  total==num:
        print(num,"Armstrong Number")
```

## Perfect Number
sum of factors of number == number

```python
num=28
sums=1
for i in range(2,int(num**0.5)):
    if num%i==0:
        sums+=i
        if i != num//i:
            sums+=num//i

if sums == num:
    print(num, "is a Perfect Number")
else:
    print(num, "is not a Perfect Number")
```

## strong number
145=1!+4!+5!

```python
start=int(input("Enter the start number: "))
ends=int(input("Enter the end number: "))

#precompute factorial values for digits 0-9
factorial=[1]*10
for i in range(1,10):
    factorial[i]=factorial[i-1]*i

for i in range(start,ends+1):
    num=i
    temp=num
    sums=0
    while temp>0:
        digit=temp%10
        sums+=factorial[digit]
        temp//=10

    if sums == num:
        print(num, "is a Strong Number")
    else:
        print(num, "is not a Strong Number")

## time complexity O(n*log(n))
```

#Spy number
1124
1+1+2+4 = 8 = 1*1*2*4

```python
## basic approach
num=1124
digit_sums=sum(int(d) for d in str(num))
digit_multiplication=1
for d in str(num):
    digit_multiplication=digit_multiplication*int(d)

if digit_sums==digit_multiplication:
    print(num, "is a Spy Number")
else:
    print(num, "is not a Spy Number")
    

## optimized approach
digit_sums=0
sdigit_multiplication=1
temp=num
while temp>0:
    digit=temp%10
    digit_sums+=digit
    sdigit_multiplication*=digit
    temp//=10
if digit_sums==sdigit_multiplication:
    print(num, "is a Spy Number")
else:
    print(num, "is not a Spy Number")
```

## Happy number
19
1^2+9^2=82
8^2+2^2=68
6^2+8^2=100
1^2+0^2+0^2=1

when sum of sqares of digits produce 1 


```python
n = int(input("Enter number: "))

seen = set()

while n != 1 and n not in seen:
    seen.add(n)

    total = 0
    temp = n

    while temp > 0:
        digit = temp % 10
        total += digit * digit
        temp //= 10

    n = total

if n == 1:
    print("Happy Number")
else:
    print("Not a Happy Number")
```

## Duck Number
if a number atleast contain 0 except leading digit

```python
num=1024
temp=num
is_duck=False
while temp >= 10:
    digit = temp % 10
    if digit==0:
        is_duck=True
        break
    temp //= 10

if is_duck:
    print(num, "is a Duck Number")
else:
    print(num, "is not a Duck Number")
```

## Magic number
19
1+9=10
1+0=1

```python
num=19
temp=num
seen=set()

while temp!=1  and not temp in seen:
    seen.add(temp)
    sums=0
    while temp>0:
        digit=temp%10
        sums+=digit
        temp//=10
    temp=sums

if sums==1:
    print(num, "is a Magic Number")
```
