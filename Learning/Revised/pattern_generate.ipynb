# Pattern Generate With techniques
 - how many rows
 - how many spaces
 - how many symbols
Then create a formula and generate pattern

## increasing pattern or Increasing triangle.


```python
# row = symbols
n=4
for i in range(1 ,n+1):
    print("*" * i)
```

## Decreasing pattern or Decreasing triangle.

```python
#
n=4
for i in range (n,0,-1):
    print("*" * i)
```

## Pyramid

```python
#space=n-row  and stars=2*i-1

for i in range (1,n+1):
    print(" " * (n-i)+"*"*(2*i-1))
```

## inversed pyramid


```python
for i in range (n,0,-1):
    print(" " * (n-i)+"*"*(2*i-1))
```

## Diamond
upper pyramid +lower inverse pyramid

```python
#upper
for i in range (1,n+1):
    print(" "*(n-i)+"*"*(2*i-1))
#lower invers
for q in range (n,0,-1):
    print(" "*(n-q)+"*"*(2*q-1))
```

## square pattern

```python
# row=column
n=10
for i in range(n):
    print("*"*n)
```

## Rectangles

```python
## column=length
## row =width
column=8
row=5
for i  in range(row):
        print("*"*column)
```

## Right Align Tringle
space=n-i
star=i

```python
n=5
for i in range(1,n+1):
    print(" "*(n-i)+"*"*i)
```

## Hollow Square

```python
# row=0 and row=n-1 or column=0 and column=n-1
n=5
for i in range(n):
    for j in range(n):
        if i==0 or i==n-1 or j==0 or j==n-1:
           print("*",end="")
        else:
            print(" ",end="")
    print()
        
```

## hollow tringle

```python
n=5
for i in range(n):
    for j in range(n):
        if j==0 or i==n-1 or i==j:
            print("*",end="")
        else:
            print(" ",end="")
    print()
```

## Number Tringle

```python
n=7
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

## continuous number tringle

```python
n=4
k=0
for i in range(1,n+1):
    for j in range(1,i+1):
        k+=1
        print(k,end=" ")
    print()
```

## alphabet tringle

```python
char=65
for i in range(n):
    for j in range(i+1):
      print(chr(char),end="")
      char+=1
    print()
```

## fibonacci tringle

```python
a=0
b=1
for i in range(n):
    for j in range(i+1):
        print(a,end=" ")
        a,b=b,a+b
    print()
```

## binary tringle

```python
for i in range(n):
    for j in range(i+1):
        if (i+j) % 2==0 :
            print("1",end="")
        else:
            print("0",end="")
    print()        
```
