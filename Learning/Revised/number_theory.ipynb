## Prime Check

```python
def prime_check(num):
     if num < 2 :
         return False
     
     for i in range(2,int(num**0.5)+1):
              if num%i==0:
                return False
                break
     return True

print(prime_check(int(input("Enter a number to check if it is prime or not: "))))
```

## Divisors

```python
num=10
divisors=[]
for i in range(1,int(num**0.5)+1):
    if num%i==0:
        divisors.append(i)
        if i!=num//i:
            divisors.append(num//i)
divisors.sort()
print(divisors)
```

## GCD,LCM

```python
## GCD Euclidean Approach
a=10
b=12
x=a;y=b
while y!=0:
 x,y=y,x%y

print("GCD: ",x)

# built in function
import math
print(f"GCD of {a} and {b} is {math.gcd(x,y)}")
```

## GCD of multiple numbers

```python
## euclidean
def gcd_finders(a,b):
     while b!=0:
       a,b=b,a%b
     return a

num=[12,10,60]
result=num[0]
for n in num[1:]:
    result=gcd_finders(result,n)
print(result)

##built in
import math
print(math.gcd(20,30,60))


## gcd of sets values
nums={10,20,30}
print(math.gcd(*nums))
```

```python
## LCD Euclidean Approach
a=10
b=12
x=a;y=b
while y!=0:
 x,y=y,x%y

gcd=x
lcm=abs(a*b)//gcd

print("LCM: ",lcm)

# built in function
import math
print(f"GCD of {a} and {b} is {math.lcm(a,b)}")
```

## Perfect/abundant/deficient classification.
perfect when sum of divisors = num
abudant when sum of divisors > num
deficient when sum of divisors < num

```python
def check_p_a_d(n):
    if n<=1:
        print(f"{n} is deficient")
        return

    sums=1
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            sums+=i

            if i!= n//i:
                sums+=n//i

    if sums==n:
        print(f"{n} is perfet")
    elif sums > n:
        print(f"{n} is abundant")
    else:
        print(f"{n} is deficient")

numbers=[6,14,28,40,65]
for i in numbers:
    check_p_a_d(i)
```
