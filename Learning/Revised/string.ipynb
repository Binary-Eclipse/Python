## Validate Password

````python
def valid_password(password):
    
    if len(password) < 8:
        return False
    if not any(ch.isupper() for ch in password):
        return False
    if not any (ch.islower() for ch in password):
        return False
    if not any (ch.isdigit() for ch in password):
            return False
    special_chars = set("!@#$%^&*()-+?_=,<>/")
    if not any(char in special_chars for char in password):
         return False

    return True


password="aaBc123@"
x=valid_password(password)

if x:
  print("Valid")
else:
    print("Invalid")
````

## Caesar Cipher
 ### A Caesar Cipher is a simple encryption technique where every letter is shifted by a fixed number of positions in the alphabet.

````python
text="ahAtcvopsk%"
shift=3
res=""
for ch in text:
    if 'a'<=ch<='z':
        distance=ord(ch)-ord('a')
        new_pos=(distance+shift)%26    # new_chr=(((ord(ch)-ord('a'))+shift)%26+ord('a')) 
        res+=chr(new_pos+ord('a'))     # res+=new_chr

    elif 'A'<=ch<='Z':
        distance=ord(ch)-ord('A')
        new_pos=(distance+shift)%26    # new_chr=(((ord(ch)-ord('a'))+shift)%26+ord('a')) 
        res+=chr(new_pos+ord('A'))     # res+=new_chr

    else:
        res+=ch
print(res)
````

# Why Are Strings Immutable in Python?

A **string is immutable** in Python. This means that once a string is created, its **individual characters cannot be changed directly**.

## Example

```python
text = "hello"

text[0] = "H"
```

This produces an error:

```text
TypeError: 'str' object does not support item assignment
```

We cannot modify the existing string `"hello"`.

Instead, we create a **new string**:

```python
text = "hello"

text = "H" + text[1:]

print(text)
```

**Output:**

```text
Hello
```

The original `"hello"` was not modified. A new string `"Hello"` was created.

---

## Why Are Strings Immutable?

### 1. Safety and Predictability

Strings cannot be changed unexpectedly after they are created.

```python
text = "hello"

new_text = text.upper()

print(text)
print(new_text)
```

**Output:**

```text
hello
HELLO
```

`upper()` does not modify `text`. Instead, it creates and returns a **new string**.

This makes strings predictable: once a string has a particular value, its contents cannot be changed by another part of the program.

---

### 2. Strings Can Be Used as Dictionary Keys

Dictionary keys must be **hashable and stable**.

Because strings are immutable, they can safely be used as dictionary keys:

```python
student = {
    "name": "Shakil",
    "age": 25
}
```

Here:

```text
"name"
"age"
```

are strings used as dictionary keys.

If strings could change after being used as keys, their hash values could change and dictionary lookups could become unreliable.

Therefore, **immutability makes strings safe to use as dictionary keys**.

---

### 3. Memory Optimization

Because strings cannot be modified, Python can sometimes **reuse the same string object** instead of storing multiple identical copies.

For example:

```python
a = "hello"
b = "hello"
```

Python may reuse the same `"hello"` object:

```text
          "hello"
          /     \
         a       b
```

This is safe because neither `a` nor `b` can modify the string.

If we write:

```python
a = "Hi"
```

Python does not change `"hello"`.

Instead, `a` now refers to another string:

```text
"hello"  ← b

"Hi"     ← a
```

This can help reduce unnecessary memory usage in situations where Python can safely reuse strings.

> **Note:** Python's string interning and object reuse are implementation details, so you should not assume that every identical string will always share the same object.

---

## What Actually Happens When We "Change" a String?

Consider:

```python
text = "hello"
text = "world"
```

Python does **not** change:

```text
"hello" → "world"
```

Instead, the variable `text` is reassigned:

```text
Before:

text ──→ "hello"


After:

text ──→ "world"
```

The original string `"hello"` itself was never modified.

## String vs List

### String — Immutable

```python
text = "abc"

text[0] = "x"      # ❌ Error
```

### List — Mutable

```python
items = ["a", "b", "c"]

items[0] = "x"     # ✅ Works
```

| Type   | Mutable? | Can change individual elements? |
| ------ | -------- | ------------------------------- |
| `str`  | ❌ No     | ❌ No                            |
| `list` | ✅ Yes    | ✅ Yes                           |

---

## In Short

**String immutability** means:

> **Once a string is created, its contents cannot be changed directly.**

Instead of modifying the existing string, Python creates a **new string**.

### Main benefits

1. **Safety and predictability** — strings cannot be unexpectedly modified.
2. **Dictionary keys** — immutable strings are stable and hashable.
3. **Memory optimization** — Python can sometimes safely reuse identical string objects.

### Remember

```text
String → Immutable → Cannot modify directly
List   → Mutable   → Can modify directly
```
