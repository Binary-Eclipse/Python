## Reverse String

```python
text= "Hello"

# slicing method
reversed_text=text[::-1]
print(reversed_text)

# reversed()

reversed_text2="".join(reversed(text))
print(reversed_text2)

# loop
reversed_text3="" #empty string
for char in text:
    reversed_text3=char+reversed_text3

print(reversed_text3)
```

## Largest Word

```python
Word="i have to do something"
splitted=Word.split()
largest=""
for i in splitted:
    if len(i)>len(largest):
          largest=i

print(largest)


#using max()
long=max(splitted,key=len)
print(long)
```

```python
#using sorted()
longest=sorted(splitted,key=len)[-1] #without key=len, Python sorts alphabetically
print(longest)
```

## shortest word

```python
Word="i have to do something"
splitted=Word.split()
shortest=splitted[0]
for i in splitted:
    if len(i)< len(shortest):
         shortest=i

print(shortest)

#using sorted()
shortestt=sorted(splitted,key=len)[0]
print(shortestt)

#using max()
short=min(splitted,key=len)
print(short)
```

## capitalize characters

```python
##using upper()
text="shy i am"
print(text.upper())

##manually
resultss=""
for char in text:
    if "a"<=char<="z":# space not allowed
        char=chr(ord(char)-32)

    resultss+=char

print(resultss)
```

**Note:**<br>
char=chr(ord(char)-32)<br>
here, **ord**: **ordinal** that convert characters into its unicode<br>
ord() → character → number <br>
and<br>
chr() → number → character

## check anagram<br>
if in two string contain same characters with same frequency<br>
listen - silent<br>
dusty - study<br>
below - elbow<br>
save - vase<br>

```python
## sorted()
text1="listen"
text2="silent"

if sorted(text1)==sorted(text2):
    print("anagram")
else:
    print("Not anagram")


## using dict
freq1={}
freq2={}

for ch in text1:
    freq1[ch]=freq1.get(ch,0)+1

for i in text2:
    freq2[i]=freq2.get(i,0)+1

if freq1==freq2:
    print("Anagram")
else:
    print("Not anagram")
```
