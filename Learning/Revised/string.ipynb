## Reverse String

```python
text= "Hello"

# slicing method
reversed_text=text[::-1]
print(reversed_text)

# reversed()

reversed_text2="".join(reversed(text))
print(reversed_text2)

# loop
reversed_text3=""
for char in text:
    reversed_text3=char+reversed_text3

print(reversed_text3)
```

## First Non-repeating character

```python
toxt="programming"
for ip in toxt:
    if toxt.count(ip)==1:
        print("first non-repeating character:",ip)
        break

# without count()
frequency={}
for char in toxt:
    frequency[char]=frequency.get(char,0)+1

for i in toxt:
    if frequency[i]==1:
        print("First non-repeating chararter: ",i)
        break;

    
```

## Largest Word

```python
Word="i have to do something"
splitted=Word.split()
largest=""
for i in splitted:
    if len(i)>len(largest):
          largest=i

print(largest)

#using sorted()
longest=sorted(splitted,key=len)[-1]
print(longest)

#using max()
long=max(splitted,key=len)
print(long)
```

## shortest word

```python
Word="i have to do something"
splitted=Word.split()
shortest=splitted[0]
for i in splitted:
    if len(i)< len(shortest):
         shortest=i

print(shortest)

#using sorted()
shortestt=sorted(splitted,key=len)[0]
print(shortestt)

#using max()
short=min(splitted,key=len)
print(short)
```

## capitalize characters

```python
##using upper()
text="shy i am"
print(text.upper())

##manually
resultss=""
for char in text:
    if "a"<=char<="z":
        char=chr(ord(char)-32)

    resultss+=char

print(resultss)
```

## capitalize word

```python
docs="hello i am shakil"
words=docs.split()
r=""
for ch in words:
    first=ch[0]

    if "a"<=first<="z":
        first=chr(ord(first)-32)
    
    r+=first+ch[1:]+" "

print(r)
```

## reversed word in a sentence

```python
text="Hello i took a pen"
sp=text.split()
result=""

for ch in sp:
    result+=ch[::-1]+" "

print(result)


## without slicing
r=""
for word in sp:
    reverse=""
    for ch in word:
        reverse=ch+reverse

    r+=reverse+" "
print(r)
```

## check anagram
if in two string contain same characters with same frequency
listen - silent
dusty - study
below - elbow
save - vase

```python
## sorted()
text1="listen"
text2="silent"

if sorted(text1)==sorted(text2):
    print("anagram")
else:
    print("Not anagram")


## using dict
freq1={}
freq2={}

for ch in text1:
    freq1[ch]=freq1.get(ch,0)+1

for i in text2:
    freq2[i]=freq2.get(i,0)+1

if freq1==freq2:
    print("Anagram")
else:
    print("Not anagram")
```
