## Convert seconds into hours, minutes, and seconds. 

```python
timeInSec=int(input("Enter Time in Second: "))

hour=timeInSec//60
rem=timeInSec%60
minute=rem//60
sec=rem%60

print(f"{hour}:{minute}:{sec}")
```

## basic calculator

```python
num1=int(input("Value1: "))
operator=input("+,-,*,/ ")
num2=int(input("Value2: "))

match operator:
    case "+" :print(num1+num2)
    case "-" :
        print(num1-num2)
    case "*" :
              print(num1*num2)
    case "/" :
        print(num1/num2)
    case _:
        print("Invalid Opertor")

    
```
